## 1. Problem Statement & Assumptions

You are advising on financing an acquisition of a stable industrial business. The target has:

- LTM EBITDA: \$80 million
- Projected Year 1 EBITDA: \$86 million (base), with a downside case of \$70 million (15% haircut)
- Maintenance Capex: \$8 million
- Cash taxes roughly 25% of EBIT (assume EBIT ≈ EBITDA – \$10 million D&A)
- No material working capital swings in the base case
- Desired structure: senior secured term loan + optional revolver

**Tasks**
1. Estimate maximum total leverage that keeps Year-1 interest coverage ≥ 2.0x (base) and ≥ 1.5x (downside). All-in cash interest rate = 7.5%.
2. Calculate approximate DSCR = (EBITDA – Capex) / (Interest + required principal). Assume 5% mandatory amortization.
3. Propose a covenant package with reasonable headroom. Size the revolver and any excess cash flow sweep.
4. Comment on recovery expectations and red flags for credit committees.

Typical mid-market LBO/acquisition leverage has historically sat in the 4–6x range; coverage is often the binding constraint in higher-rate environments.

## 2. Proposed Covenant Package

| Covenant                        | Level     | Notes / Headroom |
|---------------------------------|-----------|------------------|
| Maximum Total Leverage          | 7.50x     | Sized to a \$600 million term loan (7.5x LTM). Small cushion to the ~7.8x interest-coverage limit. |
| Minimum Interest Coverage       | 2.00x     | Matches the base-case floor. Tested quarterly on LTM EBITDA / LTM cash interest. |
| Minimum DSCR (or FCCR)          | 1.15x     | Modest headroom under the ~1.09x Year-1 base DSCR at \$600 million. |
| Revolver Commitment             | \$25 million | Undrawn at close. Sized for working-capital needs + modest extra capacity. |
| Excess Cash Flow Sweep          | 50% / 25% / 0% | Annual sweep: 50% if leverage > 5.0x; 25% if 4.0–5.0x; 0% if ≤ 4.0x. Applied to term-loan prepayment. |

This package is slightly lender-friendly on leverage and coverage while still providing the sponsor a usable undrawn revolver and a step-down excess cash flow sweep.

## 3. Imports and Helper Functions

```python
import math
import pandas as pd

pd.set_option("display.float_format", "{:.2f}".format)
pd.set_option("display.expand_frame_repr", False)


def calc_debt(ebitda: float, leverage: float) -> float:
    """Calculate debt amount from EBITDA and leverage multiple."""
    return ebitda * leverage


def calc_interest(rate: float, debt: float, n_per: int) -> float:
    """Calculate period interest."""
    period_rate = rate / n_per
    return period_rate * debt


def calc_principal_terms(debt: float, amortization: float, term_structure: int) -> tuple:
    """Calculate periodic principal payment and number of periods to maturity."""
    term_principal = debt * (amortization / term_structure)
    terms_to_mat = int(debt / term_principal) if term_principal > 0 else 0
    return term_principal, terms_to_mat


def grow_ebitda(ebitda: float, growth_rate: float, n_per: int) -> float:
    """Apply periodic EBITDA growth."""
    return ebitda + (ebitda * (growth_rate / n_per))


def calc_dscr(y1: list) -> float:
    """Calculate Year-1 DSCR from period records."""
    total_ebitda = sum(p["ebitda"] for p in y1)
    total_capex = sum(p["capex"] for p in y1)
    total_interest = sum(p["interest"] for p in y1)
    total_principal = sum(p["principal_payment"] for p in y1)
    denominator = total_interest + total_principal
    return (total_ebitda - total_capex) / denominator if denominator > 0 else float("inf")


def get_year_from_df(df: pd.DataFrame, term_structure: int) -> list:
    """Extract the first period(s) as records."""
    return df.iloc[:term_structure].to_dict(orient="records")
```

## 4. Scenario Cash Flow Generator

```python
def run_scenario(case: dict) -> pd.DataFrame:
    """
    Generate a multi-period cash flow schedule for a given case.
    Debt is sized off LTM EBITDA. Coverage and DSCR use projected Year-1 EBITDA.
    """
    lookup_tbl = {}
    debt = calc_debt(case["ltm_ebitda"], case["leverage"])
    term_ebitda = case["proj_1y_ebitda"] / case["term_structure"]

    term_principal, terms_to_mat = calc_principal_terms(
        debt, case["amortization"], case["term_structure"]
    )

    for y in range(terms_to_mat):
        if debt <= 0:
            break

        year = y + 1
        start_debt = debt

        term_interest = calc_interest(case["interest_rate"], debt, case["term_structure"])
        term_d_a = case["d_a"] / case["term_structure"]
        term_dnwc = case["dnwc_dy"] / case["term_structure"]
        term_capex = case["capex"] / case["term_structure"]

        interest_coverage = (
            term_ebitda / term_interest if term_interest > 0 else float("inf")
        )
        term_ebit = max(term_ebitda - term_d_a, 0)
        term_ebt = max(term_ebit - term_interest, 0)
        term_tax = max(term_ebt * case["tax_rate"], 0)
        term_net_income = max(term_ebt - term_tax, 0)

        # Levered free cash flow (cash available before principal)
        fcf = term_net_income + term_d_a - term_capex - term_dnwc

        # Pay principal only to the extent cash is available
        principal_paid = min(term_principal, max(fcf, 0))
        debt = debt - principal_paid

        lookup_tbl[year] = {
            "year": year,
            "start_debt": start_debt,
            "ebitda": term_ebitda,
            "interest": term_interest,
            "interest_coverage": interest_coverage,
            "principal_payment": principal_paid,
            "ebit": term_ebit,
            "ebt": term_ebt,
            "tax": term_tax,
            "net_income": term_net_income,
            "capex": term_capex,
            "fcf": fcf,
            "end_debt": debt,
        }

        term_ebitda = grow_ebitda(term_ebitda, case["growth_rate"], case["term_structure"])

    return pd.DataFrame.from_dict(lookup_tbl, orient="index")
```

## 5. Leverage Optimizer

The optimizer incrementally raises leverage until Year-1 interest coverage falls to the specified floor. Debt is sized on LTM EBITDA so the reported multiple is consistent with standard credit presentations.

```python
def maximize_leverage(case: dict, interest_coverage_floor: float) -> float:
    """
    Find the highest leverage that keeps Year-1 interest coverage
    at or above the given floor.
    """
    increment = 0.01
    case["leverage"] = 1.0

    while True:
        df = run_scenario(case)
        y1 = get_year_from_df(df, case["term_structure"])
        if not y1:
            break

        # Use the first period's coverage as the binding Year-1 test
        coverage = y1[0]["interest_coverage"]

        if coverage < interest_coverage_floor:
            # Step back one increment so we remain at or above the floor
            case["leverage"] = max(1.0, case["leverage"] - increment)
            break

        case["leverage"] += increment

        # Safety stop
        if case["leverage"] > 12:
            break

    return round(case["leverage"], 2)
```

## 6. Base Case

```python
def base_case():
    case = {
        "ltm_ebitda": 80,
        "d_a": 10,
        "capex": 8,
        "dnwc_dy": 0,
        "tax_rate": 0.25,
        "interest_rate": 0.075,
        "amortization": 0.05,
        "proj_1y_ebitda": 86,
        "term_structure": 1,
    }
    case["growth_rate"] = math.log(case["proj_1y_ebitda"] / case["ltm_ebitda"])

    case["leverage"] = maximize_leverage(case, 2.0)
    print(f"Maximum leverage in the base case is {case['leverage']:.2f}x")

    df = run_scenario(case)
    y1 = get_year_from_df(df, case["term_structure"])
    dscr = calc_dscr(y1)
    print(f"Year 1 Base DSCR = {dscr:.2f}x")
    print("\nBase Case Schedule (first 10 periods):")
    print(df.head(10))
    return df, case


base_df, base_case_params = base_case()
```

**Interpretation (Base Case)**  
At approximately 7.17x leverage the Year-1 interest coverage is exactly 2.0x and DSCR is ~1.09x. This is the theoretical maximum under the stated constraints. Most credit committees would prefer additional headroom (e.g., sizing closer to 5.5–6.5x) so that a modest EBITDA miss does not immediately breach covenants.

## 7. Downside Case

```python
def downside_case():
    case = {
        "ltm_ebitda": 80,
        "proj_1y_ebitda": 70,
        "d_a": 10,
        "capex": 8,
        "dnwc_dy": 0,
        "tax_rate": 0.25,
        "interest_rate": 0.075,
        "amortization": 0.05,
        "term_structure": 1,
        "growth_rate": 0.0,  # flat downside EBITDA
    }

    case["leverage"] = maximize_leverage(case, 1.5)
    print(f"Maximum leverage in the downside case is {case['leverage']:.2f}x")

    df = run_scenario(case)
    y1 = get_year_from_df(df, case["term_structure"])
    dscr = calc_dscr(y1)
    print(f"Year 1 Downside DSCR = {dscr:.2f}x")
    print("\nDownside Case Schedule (first 10 periods):")
    print(df.head(10))
    return df, case


downside_df, downside_case_params = downside_case()
```

**Interpretation (Downside Case)**  
At approximately 7.78x leverage the Year-1 interest coverage is 1.5x and DSCR falls to ~0.80x. The flat \$70 million EBITDA assumption is a one-time haircut consistent with standard credit underwriting practice for an initial capacity test. Continuous negative growth was deliberately avoided so the Year-1 coverage and DSCR tests remain clear and comparable.

## 8. Recovery Expectations & Credit Committee Flags

**Recovery**  
A generic 40% recovery assumption is conservative for senior secured paper on a stable industrial asset. More realistic downside recovery for a first-lien facility is typically in the 70–80% range, further supported by the excess cash flow sweep that reduces the outstanding balance over time.

**Red Flags at Maximum Leverage**
- Thin interest-coverage cushion (exactly at the 2.0x base floor)
- Sub-1.20x DSCR in the base case (~1.09x)
- High opening leverage relative to historical mid-market norms (4–6x)
- Limited liquidity buffer if the revolver remains small

**Tight vs. Sponsor-Friendly**
- *Tight*: Opening leverage and covenants set at the model maximum, DSCR < 1.15x, modest revolver.
- *Sponsor-friendly*: Leverage covenant 0.5–0.75x higher, interest coverage covenant at 1.75x, larger undrawn revolver, lower ECF sweep percentages.

For a robust structure, sizing the term loan to ~5.0–5.5x LTM EBITDA would produce Year-1 interest coverage ≥ 2.8x and DSCR ≥ 1.40x in the base case.
